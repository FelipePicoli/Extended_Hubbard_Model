```julia
using Plots, LaTeXStrings
using Plots.Measures
using CSV
using DataFrames

# Set global defaults for a clean LaTeX look
default(
    fontfamily = "Computer Modern",
    tickfont = "Computer Modern",
    legendfont = "Computer Modern",
    guidefont = "Computer Modern",
    titlefont = "Computer Modern",
    grid = false,
    titlefontsize   = 15,  # Main title
    guidefontsize   = 15,  # x and y labels
    tickfontsize    = 15,  # Numbers on the axes
    legendfontsize  = 15,  # Legend text
    legendtitlefontsize = 15,
    linewidth       = 2,   # While we're at it, make lines thicker for the big fonts
    frame = :box,
    dpi = 200,        # Sets the resolution for display and saving
    # size = (800, 600) # Sets the physical pixel dimensions
)
```

```julia
df_1 = CSV.read("../src/result_U=0.0.csv", DataFrame)
df_2 = CSV.read("../src/result_U=-0.95.csv", DataFrame)
df_3 = CSV.read("../src/result_U=-1.76.csv", DataFrame)
df_4 = CSV.read("../src/result_U=-2.85.csv", DataFrame)
df_5 = CSV.read("../src/result_U=-7.73.csv", DataFrame)
```

```julia
x = df_1[!, "V"]
data = [df_1[!, "E_p"], df_2[!, "E_p"], df_3[!, "E_p"], df_4[!, "E_p"], df_5[!, "E_p"]]
p = plot(x, data,
     label=[L"0.0" L"-0.95" L"-1.76" L"-2.85" L"-7.73"],
     xlabel=L"$V/J$",
     ylabel=L"$E_{p}$", 
     legend = :bottomright,
     legendtitle = L"U/J",

     linestyle = [:solid :dash :dot :dashdot :dash], 
     markershape = [:circle :rect :utriangle :diamond :hexagon],
     markersize = 4,
     markerstrokewidth = 0.25,
     markerstrokecolor = :white,
     lw = 1.5,
     alpha = 0.8,
     titlefontsize   = 15,  # Main title
     title="Entanglement of particles",
     size = (400, 300))
# savefig(p, "entanglement_of_particles_m=200.pdf")
```

```julia
df = CSV.read("../src/result_U=1.6_N=32.csv", DataFrame)
x = df[!, "V"]
data = [df[!, "E_p"]]

p1 = plot(x, data,
    legendtitle = L"U/J",
    label="1.6",
    ylabel="Entanglement of particles", 
    legend = :bottomright,
    xticks = nothing,         # Remove tick values

    linestyle = [:solid :dash :dot :dashdot :dash], 
    markershape = [:circle :rect :utriangle :diamond :hexagon],
    markersize = 4,
    markerstrokewidth = 0.25,
    markerstrokecolor = :white,
    
    # ylims = (0.1, 0.250),  
    bottom_margin = -4mm,

    lw = 1.5,
    alpha = 0.8,
    xflip = true,
    
    titlefontsize   = 15,  # Main title
    # title=""
    )

p2 = plot(x, df[!, "single_site_entanglemenet"],
    legendtitle = L"U/J",
    label="1.6",
    xlabel=L"$V/J$",
    ylabel="Single site entanglement", 
    legend = :bottomleft,
    linestyle = [:solid :dash :dot :dashdot :dash], 
    markershape = [:circle :rect :utriangle :diamond :hexagon],
    markersize = 4,
    markerstrokewidth = 0.25,
    markerstrokecolor = :white,
    
    top_margin = -9mm,
    ylims = (1.90, 1.965),  
    lw = 1.5,
    alpha = 0.8,
    xflip = true,
    titlefontsize   = 15,)

p = plot(p1, p2,
    layout = grid(2, 1, heights=[0.5, 0.5]),
    size = (450, 700),
    link = :x,
    bottom_margin = 5mm,
    margin = 7.5mm) 


vline!(p[1], [-1.58], label = L"-1.55", color=:green, linestyle=:dashdot)
vline!(p[2], [-1.58], label = L"-1.55", color=:green, linestyle=:dashdot)

vline!(p[2], [-1.1], label = L"-1.1", color=:red, linestyle=:dot)
vline!(p[1], [-1.1], label = L"-1.1", color=:red, linestyle=:dot)

# savefig(p, "entanglement_particles_single_site_N=32.pdf")
```
